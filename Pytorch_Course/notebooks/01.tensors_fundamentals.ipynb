#### What is a Tensor?
Tensor is a fundamental building block of Pytorch, It is a collection of numbers(n dim array) that can be processed by GPUs in the system.

```python
# Let us create our all 4 types of TENSORS
#1. scalar
scalar = torch.tensor(3)
print(scalar)
print(scalar.shape)
print(scalar.ndim)
```

```python
#1. vector
vector = torch.tensor([3])
print(vector)
print(vector.shape)
print(vector.ndim)

vec2 = torch.tensor([3,4,5])
print(vec2.ndim)
print(vec2.shape)
```

### Random Tensors
Why Random Tensors?
- Random tensors are really important in Neural Networks because what NN does is that we start with any random tensors and then update the values of the tensors to get the better representation of the data we are working on.

```python
# How to create random tensor of size/shape (3 , 4)?
random_tensor = torch.rand((3,4))
print(random_tensor)
print(random_tensor.ndim)
print(random_tensor.shape)

# Create a random tensor which looks like a coloured image_data
random_img_data_tensor = torch.rand(size = (3,224,224)) # This gives us height = 224, width = 224, colour channels = 3
print(random_img_data_tensor.ndim)
print(random_img_data_tensor.shape)
```

```python
random_tensor = torch.randn(size=(100,)) # gives the numbers which are Normally distributed with mean=0, std=1
print(random_tensor) 
print(random_tensor.ndim)
print(random_tensor.shape)
```

```python
# use of manual seed
torch.manual_seed(100)
torch.rand(2,3)
```

```python
torch.manual_seed(100)
torch.rand(2,3) # reproducibility -> always the same random numbers are given with the same seed
```

### Tensor Shape

```python
x = torch.rand(3,4)
print(x.shape) # gives the shape of tensor (x)

# To create a tensor with the same shape as x:
print(torch.empty_like(x)) # creates an empty tensor of the same shape as tensor x
print(torch.rand_like(x)) # creates an random tensor of the same shape as tensor x
```

### Aggregate Operations/Reduction Operations


```python
e = torch.tensor([
    [2,3,4],
    [6,7,8]
] , dtype=torch.float32)
```

```python
# sum

# i. total sum
print(torch.sum(e)) # total sum of values inside the tensor x

# ii. sum along the columns
print(torch.sum(e,dim=0))

# iii. Along the rows
print(torch.sum(e , dim=1))
```

```python
# average/mean
print(torch.mean(e))

# i. along column
print(torch.mean(e , dim=0))

# ii. along Row
print(torch.mean(e , dim=1))
```

```python
# other functions like median() , min(), max(), product() , std()

# Important 
# argmax -> tells the index of the largest value in a tensor

print(torch.argmax(e)) # largest value index in e (this is after we flatten)
print(torch.argmin(e)) # lowest value index in e (this is after we flatten)

e[0]
```
